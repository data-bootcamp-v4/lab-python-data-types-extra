```python
diccionario = {}

diccionario['nombre'] = input('ingresa tu nombre')
diccionario['edad'] = int(input('ingresa tu edad'))
diccionario['direccion'] = (input('ingresa tu direccion'))
diccionario['salario'] = float(input('ingresa tu salario'))
diccionario['gasto'] = float(input('ingresa tus gasto mensual'))
```

```python
salario_restante = round(diccionario.get('salario') - diccionario.get('gasto'), 1)
salario_restante
```

```python
salario_bueno = salario_restante >= 500
```


### Exercise 2: Text Cleaning and Concatenation

Write code that takes text from the `poem` variable, removes punctuation, and leaves everything lowercase. 

Concatenate the resulting string with the string "python is awesome!" and store the result in a new variable. 

Prints the length of the new string.

Split the string into a list of strings using the space delimiter, store it in a variable `poem_list` and print it.

*Hint:*
- *You can use the len() function to get the length of a string.*
- *Find string methods to perform the other steps. Recommended external resources: [Python String Methods](https://www.w3schools.com/python/python_ref_string.asp)*
- *Use method chaining to simplify your code. If you are not sure what it is, read this tutorial first: https://pyneng.readthedocs.io/en/latest/book/04_data_structures/method_chaining.html*

```python
poem = """Some say the world will end in fire,
Some say in ice.
From what I’ve tasted of desire
I hold with those who favor fire.
But if it had to perish twice,
I think I know enough of hate
To say that for destruction ice
Is also great
And would suffice."""
```

```python
import string
poem_low = poem.lower()
poema_limpio = poem_low.translate(str.maketrans('','', string.punctuation + '’'))
```

```python
texto = "python is awesome!"
nuevo_texto = poema_limpio + " " + texto
print(len(nuevo_texto))
```

```python
poem_list = nuevo_texto.split()
print(poem_list)
```
